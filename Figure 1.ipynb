# Figure 1b - Phylogenetic tree

## Open tree file

```r
path <- "./lib/"  
treefile <- file.path(path, "ath_475ppr.pep.fa.phy.contree")
tree <- read.tree(treefile)

p <- ggtree(tree, color = "black", layout = "fan", size = 1.5) +
  geom_tiplab2(offset = 0.08, size = 0, color = "grey", align = TRUE, linetype = NA)
```

## Add chr information

```r
gene_ids <- tree$tip.label

chr_data <- data.frame(
  ID = gene_ids,
  Chromosome = factor(sub("AT([1-5])G.*", "\\1", gene_ids)),
  Count = factor(1)  
)

p <- p + new_scale_fill() +
  geom_fruit(data = chr_data, geom = geom_tile,
             mapping = aes(y = ID, x = Count, fill = Chromosome),
             color = "white", offset = 0.02, width = 0.2) +
  scale_fill_manual(values = c("1"="#9E2214", "2"="#4363d8", "3"="#DA8932", 
                               "4"="#390996", "5"="#8EF6C4"),
                    guide = guide_legend(keywidth=2, keyheight=2))


addline <- data.frame(ID = chr_data$ID, Count = factor(1))
p <- p + new_scale_fill() +
  geom_fruit(data = addline, geom = geom_tile,
             mapping = aes(y = ID, x = Count),
             color = "black", fill = "black", offset = 0.015, width = 0.02)
```

## Add sRNA information

```r
# load tpm data
tpm_data <- read.delim2("./lib/avg.tpm_srna_accs.txt", sep = "\t", header = TRUE, check.names = FALSE)

# load target infor
tar_data <- read.delim2("./lib/srna-ppr_accs.txt", sep = "\t", header = TRUE, check.names = FALSE)
tar_data <- tar_data %>% filter(AllenScore < 4) %>% mutate(Gene = sub("\\..*", "", ids))

srna_df <- tar_data %>%
  select(Query, Gene) %>%
  distinct() %>%
  mutate(value = 1) %>%
  pivot_wider(names_from = Query, values_from = value, values_fill = 0)
names(srna_df)[1] <- "Gene"

# merge tpm and target infor 
srna_df_full <- tpm_data %>%
  select(ID) %>%
  left_join(srna_df, by = c("ID"="Gene")) %>%
  replace(is.na(.), 0)

srna_agg <- srna_df_full %>%
  mutate(value = rowSums(across(-ID))) %>%
  transmute(ID, value,
            Range_Label = case_when(
              value == 0 ~ "0",
              value == 1 ~ "1",
              value %in% 2:4 ~ "2-4",
              value %in% 5:7 ~ "5-7",
              TRUE ~ "Other"),
            Count = factor(1))

# add sRNA infor
p <- p + new_scale_fill() +
  geom_fruit(data = srna_agg, geom = geom_tile,
             mapping = aes(y = ID, x = Count, fill = Range_Label),
             color = "white", offset = 0.014, width = 0.2) +
  scale_fill_manual(name = "Trigger number",
                    values = c("0"="white","1"="#F4CFDD","2-4"="#E799B6","5-7"="#BE4F81"))

p <- p + new_scale_fill() +
  geom_fruit(data = srna_agg, geom = geom_tile,
             mapping = aes(y = ID, x = Count),
             colour = "black", fill = "black", offset = 0.014, width = 0.02)
```

## Add sRNA TPM information

```r
tpm_data <- read.delim2("./lib/avg.tpm_srna_accs.txt", sep = "\t", header = TRUE, check.names = FALSE)
rownames(tpm_data) <- tpm_data$ID
tpm_data$ID <- NULL
tpm_data$Average <- NULL

tpm_data[] <- lapply(tpm_data, function(x) as.numeric(as.character(x)))
tpm_data <- replace(tpm_data, is.na(tpm_data), 0)
sRNA_tpm <- log2(tpm_data + 1)


sRNA_tpm$ID <- rownames(sRNA_tpm)
sRNA_tpm <- melt(sRNA_tpm, id.vars = "ID", variable.name = "Accession", value.name = "Abundance")
sRNA_tpm$Abundance <- ifelse(is.na(sRNA_tpm$Abundance), 0, sRNA_tpm$Abundance)


p <- p + new_scale_fill() +
  geom_fruit(data = sRNA_tpm, geom = geom_tile,
             mapping = aes(y = ID, x = Accession, alpha = Abundance, fill = Accession),
             color = "white", offset = 0.012, size = 0.2) +
  scale_fill_manual(values=c("An-1"="#0000FF","Col-0"="#FFA500","Ct-1"="#FF0000",
                             "Cvi-1"="#800000","Eri-1"="#000080","Kyo-1"="#008000",
                             "Ler-0"="#800080","Sha"="#696969"),
                    guide=guide_legend(keywidth=2, keyheight=2, order=4)) +
  scale_alpha_continuous(name="TPM (log2)", range=c(0,1),
                         guide=guide_legend(keywidth=2, keyheight=2, order=5,
                                             legend.text=element_text(size=20)))
```

## Save figure to file

```r
p <- p + theme(
  legend.position = c(0.95, 0.5),
  legend.key.size = unit(0.5, 'cm'),
  legend.background = element_rect(fill = NA),
  legend.title = element_text(size = 20),
  legend.text = element_text(size = 20),
  legend.spacing.y = unit(0.5, "cm"),
  legend.spacing.x = unit(0.5, "cm"),
  plot.margin = unit(c(5, 5, 5, 5), "mm")
)

# Save plot
ggsave("ath_475ppr.pep.fa.phy.contree.pdf",
       plot = p, width = 100, height = 100, units = "cm", limitsize = FALSE)
```

# Figure 1c - sRNA distribution on Chr1-5

```r
# load sRNA count data  
data <- read.delim2("./lib/srna_count_accs.txt", sep = " ", header = FALSE, stringsAsFactors = FALSE)
data <- data[, !is.na(data[1, ])]
names(data) <- c("Pos", "Count", "Strand", "Chr", "Acc")

data$Count <- abs(as.numeric(data$Count))
data$Strand <- "plus"

# define accession order
data <- within(data, Acc <- factor(Acc, levels = c("An-1", "Col-0", "Ct-1", "Cvi-1", "Eri-1", "Kyo-1", "Ler-0", "Sha")))
with(data, levels(Acc))

# remove chr6 and chr7 data
data <- data[!data$Chr %in% c("Chr6", "Chr7"), ]

# convert Start to Mb and log10-transform 
data$Pos <- as.numeric(data$Pos) / 1000
data$Count <- abs(log10(as.numeric(data$Count)))
```

```r
# load chr ids across accessions
cdata <- read.delim2("./lib/chr_ids_accs.txt", sep = "\t", header = TRUE, stringsAsFactors = FALSE)
for (i in 1:nrow(cdata)) {
  clist <- unique(as.character(cdata[i, ]))
  clist <- sapply(clist, function(x) gsub(" ", "", x))
  data$Chr[data$Chr %in% clist] <- gsub(" ", "", cdata[i, 1])
}
```

## Visualize coverage of sRNAs on chromosomes

```r
p <- ggplot(data, aes(x = Pos, y = Count)) +
        facet_grid(Acc ~ Chr, scales = "free_x", space = "free_x") +
        geom_bar(aes(fill = Acc), stat = "identity") +
        scale_fill_manual(name = "Accession",
            values = c(
              "An-1" = "#0000FF", "Col-0" = "#FFA500", "Ct-1" = "#FF0000", "Cvi-1" = "#800000",
              "Eri-1" = "#000080", "Kyo-1" = "#008000", "Ler-0" = "#800080", "Sha" = "#696969"
            ),
            guide = guide_legend(keywidth = 0.8, keyheight = 0.8, order = 4)
        ) +
        scale_x_continuous(expand = c(0, 0)) +
        theme(axis.text.x = element_text(size = 15, colour = "black"),
              axis.text.y = element_text(size = 15, colour = "black")) +
        xlab("Chromosome position (Mb)") +
        ylab("Coverage (log10)") +
        theme_classic() +
        theme(strip.background = element_rect(colour = "white", fill = "white"),
              legend.background = element_rect(fill = NA),
              legend.title = element_text(size = 12, face = "bold"),
              legend.text = element_text(size = 10, face = "plain"))

# Save plot
ggsave(filename = "chr_sRNA_coverage_accs.pdf", plot = p, height = 4, width = 12)
```

## Visualize coverage of sRNAs from hotspot region

```r
# load hotspot region data and extract sRNA data from hotspot region
hotspot <- read.table("./lib/hotspot_range.txt", header = TRUE, stringsAsFactors = FALSE)
hotspot_data <- data %>%
  left_join(hotspot, by = c("Chr" = "ChrID", "Acc" = "Accession")) %>%
  filter(Pos >= 21 & Pos <= 24)   
```

```r
p <- ggplot(hotspot_data, aes(x = Pos, y = Count, fill = Acc)) +
  facet_grid(rows = vars(Acc), scales = "free_x", space = "free_x") +
  geom_bar(stat = "identity", width = 0.03, alpha = 0.6) +
  scale_fill_manual(
    name = "Accession",
    values = c(
      "An-1" = "#0000FF", "Col-0" = "#FFA500", "Ct-1" = "#FF0000", "Cvi-1" = "#800000",
      "Eri-1" = "#000080", "Kyo-1" = "#008000", "Ler-0" = "#800080", "Sha" = "#696969"
    ),
    guide = guide_legend(keywidth = 0.8, keyheight = 0.8, order = 4)
  ) +
  scale_x_continuous(expand = c(0, 0)) +
  xlab("Hotspot position (Mb)") +
  ylab("Coverage (log10)") +
  theme_classic() +
  theme(
    strip.background = element_rect(colour = "white", fill = "white"),
    axis.text.x = element_text(size = 8, colour = "black"),
    axis.text.y = element_text(size = 8, colour = "black"),
    axis.title = element_text(size = 10),
    legend.background = element_rect(fill = NA),
    legend.title = element_text(size = 8),
    legend.text = element_text(size = 8)
  )

# Save plot
ggsave(filename = "hotspot_sRNA_coverage_accs.pdf", plot = p, height = 4, width = 3.5)
```
